### Summary

This project followed a complete end-to-end Natural Language Processing (NLP) and machine learning workflow, starting from data understanding and cleaning and progressing toward text classification using machine learning. The first stage involved understanding the dataset through basic data analysis techniques such as `head()`, `tail()`, `shape`, `info()`, `describe()`, and value-count analysis. This helped identify the structure of the dataset, understand the variables, examine data types, and determine the distribution of the target classes. Data quality was then assessed by checking for missing values, duplicate records, inconsistent entries, undefined values, corrupted data, and inappropriate data types. These issues were handled appropriately to improve the reliability and consistency of the dataset.

Exploratory Data Analysis (EDA) was performed to understand patterns and relationships within the data. Univariate analysis was used to study individual variables, while bivariate and multivariate analysis helped identify relationships between variables and the target. Visualizations using Matplotlib and Seaborn provided a clearer understanding of class distributions, trends, and important characteristics of the dataset. This stage was important because it helped identify potential data-quality issues and guided subsequent preprocessing and feature-engineering decisions.

Since the main data consisted of textual information, extensive text preprocessing was performed. This included expanding contractions, converting text to lowercase, removing punctuation, URLs, unnecessary words and digits, stopwords, and extra whitespace. Tokenization was used to divide text into individual meaningful units. Text normalization was applied to bring words into a consistent form, while techniques such as Part-of-Speech (POS) tagging were considered where relevant. These preprocessing steps reduced noise and produced cleaner text suitable for machine learning.

After preprocessing, the textual data had to be converted into numerical form because machine learning algorithms cannot directly process raw text. **TF-IDF (Term Frequency–Inverse Document Frequency)** was used for text vectorization. It assigns higher importance to words that are informative within a document but less common across the entire corpus. Unigrams and bigrams can also capture individual words and meaningful short phrases.

Feature manipulation and feature selection were then performed to reduce dimensionality and retain the most relevant information. **Chi-Square feature selection** was selected because it is well suited for text classification and sparse TF-IDF matrices. Features with higher Chi-Square scores have stronger associations with the target classes and are therefore more useful for classification. Other possible techniques include Mutual Information, ANOVA, RFE, and L1-based selection.

Additional scaling was not required because TF-IDF already produces normalized numerical features and the resulting matrix is sparse and high-dimensional. Class distribution was also examined to determine whether the dataset suffered from class imbalance. Where necessary, class weighting could be applied instead of blindly using oversampling techniques such as SMOTE, which can be less suitable for high-dimensional sparse text data.

Finally, the data was divided into training and testing sets using a **stratified split** to maintain class proportions. **Logistic Regression** was implemented as the first machine-learning model because it performs well with high-dimensional sparse TF-IDF features, is computationally efficient, and provides an interpretable baseline. Model performance was evaluated using **accuracy, precision, recall, F1-score, classification report, and confusion matrix**.

Overall, the project demonstrates a complete NLP pipeline: **Data Understanding → Data Cleaning → EDA → Text Preprocessing → TF-IDF Vectorization → Feature Selection → Class Imbalance Analysis → Train-Test Split → Model Training → Evaluation**. The project also establishes a strong foundation for future improvements using SVM, Naive Bayes, ensemble models, hyperparameter tuning, cross-validation, and advanced transformer-based NLP models.

## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# Write your code to make your dataset analysis ready.
# ============================================================
# DATA WRANGLING - MAKE DATASET ANALYSIS READY
# ============================================================

import pandas as pd
import numpy as np

# Create a copy of the original dataset
credit_df = df.copy()

# -------------------------------
# Remove Duplicate Records
# -------------------------------
credit_df.drop_duplicates(inplace=True)

# -------------------------------
# Remove Leading & Trailing Spaces
# -------------------------------
object_cols = credit_df.select_dtypes(include='object').columns

for col in object_cols:
    credit_df[col] = credit_df[col].astype(str).str.strip()

# -------------------------------
# Replace Invalid Values
# -------------------------------
credit_df.replace(
    ['_', '__10000__', '_______', 'NA', 'N/A', 'NULL', 'null', 'nan', 'NaN'],
    np.nan,
    inplace=True
)

# -------------------------------
# Convert Numerical Columns
# -------------------------------
numeric_columns = [
    'Age',
    'Annual_Income',
    'Monthly_Inhand_Salary',
    'Num_Bank_Accounts',
    'Num_Credit_Card',
    'Interest_Rate',
    'Num_of_Loan',
    'Delay_from_due_date',
    'Num_of_Delayed_Payment',
    'Changed_Credit_Limit',
    'Num_Credit_Inquiries',
    'Outstanding_Debt',
    'Credit_Utilization_Ratio',
    'Total_EMI_per_month',
    'Amount_invested_monthly',
    'Monthly_Balance'
]

for col in numeric_columns:
    if col in credit_df.columns:
        credit_df[col] = pd.to_numeric(credit_df[col], errors='coerce')

# -------------------------------
# Handle Missing Values
# -------------------------------

# Numerical Columns → Median
num_cols = credit_df.select_dtypes(include=['int64', 'float64']).columns

for col in num_cols:
    credit_df[col] = credit_df[col].fillna(credit_df[col].median())

# Categorical Columns → Mode
cat_cols = credit_df.select_dtypes(include='object').columns

for col in cat_cols:
    credit_df[col] = credit_df[col].fillna(credit_df[col].mode()[0])

# -------------------------------
# Final Dataset Summary
# -------------------------------
print("=" * 60)
print("        DATASET READY FOR ANALYSIS")
print("=" * 60)

print(f"Dataset Shape        : {credit_df.shape}")
print(f"Duplicate Rows       : {credit_df.duplicated().sum()}")
print(f"Missing Values       : {credit_df.isnull().sum().sum()}")
print(f"Number of Variables  : {credit_df.shape[1]}")
print(f"Number of Records    : {credit_df.shape[0]}")

print("\nData Types:")
display(credit_df.dtypes)

print("\nFirst Five Records:")
display(credit_df.head())


````

### 2. Handling Outliers

````python
# Handling Outliers & Outlier treatments
import pandas as pd
import numpy as np

# --------------------------------------------------
# STEP 1: Select numerical variables
# --------------------------------------------------

outlier_columns = [
    'Age',
    'Annual_Income',
    'Monthly_Inhand_Salary',
    'Num_Bank_Accounts',
    'Num_Credit_Card',
    'Interest_Rate',
    'Num_of_Loan',
    'Delay_from_due_date',
    'Num_of_Delayed_Payment',
    'Changed_Credit_Limit',
    'Num_Credit_Inquiries',
    'Outstanding_Debt',
    'Credit_Utilization_Ratio',
    'Total_EMI_per_month',
    'Amount_invested_monthly',
    'Monthly_Balance'
]


# --------------------------------------------------
# STEP 2: Detect outliers using IQR
# --------------------------------------------------

outlier_summary = []

for col in outlier_columns:

    Q1 = credit_df[col].quantile(0.25)
    Q3 = credit_df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = credit_df[
        (credit_df[col] < lower_bound) |
        (credit_df[col] > upper_bound)
    ]

    outlier_summary.append({
        'Variable': col,
        'Q1': Q1,
        'Q3': Q3,
        'IQR': IQR,
        'Lower Bound': lower_bound,
        'Upper Bound': upper_bound,
        'Outlier Count': len(outliers)
    })


outlier_summary_df = pd.DataFrame(outlier_summary)

display(outlier_summary_df)
````

We chose capping instead of deleting rows because extreme financial values such as high income, debt, EMI, or credit inquiries can be legitimate customer values. Removing those customers could result in loss of useful information.

### 4. Textual Data Preprocessing

#### 1. Expand Contraction

````python
# Expand Contraction
# ---------------------------------------------
# EXPAND CONTRACTIONS
# ---------------------------------------------

import re

contractions = {
    "can't": "cannot",
    "won't": "will not",
    "n't": " not",
    "'re": " are",
    "'ve": " have",
    "'ll": " will",
    "'d": " would",
    "'m": " am",
    "'s": " is",
}

def expand_contractions(text):
    if pd.isna(text):
        return text

    text = str(text)

    for contraction, expansion in contractions.items():
        text = re.sub(
            contraction,
            expansion,
            text,
            flags=re.IGNORECASE
        )

    return text


# Apply to textual columns
text_columns = credit_df.select_dtypes(
    include='object'
).columns

for col in text_columns:
    credit_df[col] = credit_df[col].apply(
        expand_contractions
    )

print("Contraction expansion completed.")
````

#### 2. Lower Casing

````python
# Lower Casing
# Convert all textual columns to lowercase

text_columns = credit_df.select_dtypes(include='object').columns

for col in text_columns:
    credit_df[col] = credit_df[col].str.lower()

print("Lower casing completed.")
````

#### 3. Removing Punctuations

````python
# Remove Punctuations
import string

text_columns = credit_df.select_dtypes(include='object').columns

for col in text_columns:
    credit_df[col] = credit_df[col].apply(
        lambda x: x.translate(str.maketrans('', '', string.punctuation))
        if pd.notna(x) else x
    )

print("Punctuation removal completed.")
````

#### 4. Removing URLs & Removing words and digits contain digits.

````python
import re

text_columns = credit_df.select_dtypes(include='object').columns

for col in text_columns:
    credit_df[col] = credit_df[col].apply(
        lambda x: re.sub(r'https?://\S+|www\.\S+', '', str(x))
        if pd.notna(x) else x
    )

print("URLs removed successfully.")
````

#### 5. Removing Stopwords & Removing White spaces

````python
# Remove Stopwords
import re
from sklearn.feature_extraction.text import ENGLISH_STOP_WORDS

def remove_stopwords(text):
    if pd.isna(text):
        return text

    words = str(text).split()
    words = [word for word in words if word.lower() not in ENGLISH_STOP_WORDS]

    return ' '.join(words)

text_columns = credit_df.select_dtypes(include='object').columns

for col in text_columns:
    credit_df[col] = credit_df[col].apply(remove_stopwords)

print("Stopwords removed successfully.")
````

````python
# Remove White spaces
import re

text_columns = credit_df.select_dtypes(include='object').columns

for col in text_columns:
    credit_df[col] = credit_df[col].apply(
        lambda x: re.sub(r'\s+', ' ', str(x)).strip()
        if pd.notna(x) else x
    )

print("Extra white spaces removed successfully.")
````

````python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
import re # Import re for regex operations

# Fix: Ensure credit_df is defined. NOTE: If previous text preprocessing steps
# were applied to 'credit_df', initializing it here will reset those changes.
# For a complete and correct workflow, ensure all prior data wrangling and
# text preprocessing cells are executed before this cell.
if 'credit_df' not in globals():
    credit_df = df.copy()

# Assuming 'Type_of_Loan' is the textual column intended for vectorization
# After previous steps, 'Type_of_Loan' likely contains strings like "['loan1', 'loan2']"
# This function cleans those strings for TF-IDF.
def clean_list_string_for_tfidf(text):
    if pd.isna(text) or not isinstance(text, str):
        return '' # Return empty string for NaN or non-string values
    # Remove brackets, single quotes, double quotes, and commas
    cleaned_text = re.sub(r"(\[|\]|'|\"|,)", "", text)
    # Normalize spaces (replace multiple spaces with single space and strip leading/trailing)
    cleaned_text = re.sub(r'\s+', ' ', cleaned_text).strip()
    return cleaned_text

# Apply this cleaning to the Type_of_Loan column to prepare it for TF-IDF
credit_df['Type_of_Loan_cleaned'] = credit_df['Type_of_Loan'].apply(clean_list_string_for_tfidf)

# Initialize TF-IDF Vectorizer
tfidf = TfidfVectorizer(
    max_features=5000,
    ngram_range=(1, 2),
    min_df=2,
    max_df=0.95
)

# Fit and transform the cleaned text data
X = tfidf.fit_transform(credit_df['Type_of_Loan_cleaned'])

# Check the shape of the vectorized data
print("Shape of TF-IDF matrix:", X.shape)
````

#### 2. Feature Selection

````python
# Select your features wisely to avoid overfitting
from sklearn.feature_selection import SelectKBest, chi2

# Define the target variable 'y'
y = credit_df['Credit_Score']

# Select the top 3000 features using Chi-Square test
selector = SelectKBest(score_func=chi2, k=3000)

# Fit the selector and transform the TF-IDF features
X_selected = selector.fit_transform(X, y)

print("Original shape:", X.shape)
print("Shape after feature selection:", X_selected.shape)

#Find the Important Features
# Get the Chi-Square scores of all features
chi2_scores = selector.scores_

# Get feature names
feature_names = tfidf.get_feature_names_out()

# Create a DataFrame containing feature names and their scores
feature_scores = pd.DataFrame({
    'Feature': feature_names,
    'Chi2_Score': chi2_scores
})

# Sort features according to their Chi-Square score
feature_scores = feature_scores.sort_values(
    by='Chi2_Score',
    ascending=False
)

# Display the top 20 important features
feature_scores.head(20)

#Visualize the Most Important Features
import matplotlib.pyplot as plt

top_features = feature_scores.head(20)

plt.figure(figsize=(10, 6))
plt.barh(
    top_features['Feature'][::-1],
    top_features['Chi2_Score'][::-1]
)

plt.xlabel("Chi-Square Score")
plt.ylabel("Features")
plt.title("Top 20 Important Features")
plt.show()
````
